## Step 2. Create the conda environment

Run **one** of the cells below to create the environment. Mamba is recommended as it resolves dependencies much faster than conda.

This will:
1. Create a new environment called **`waporipa`**
2. Install Python 3.10 and all required packages
3. Resolve and install all transitive dependencies

````python
env_file = os.path.join(os.path.split(os.getcwd())[0], 'environment.yml')

# Option A: Using mamba (faster, recommended)
!mamba env create -f {env_file}

# Option B: Using conda (uncomment the line below and comment out Option A)
# !conda env create -f {env_file}
````

## Step 3. Use the new environment

The environment is now created and ready to use. When you open any of the module notebooks (Module 1a, 1b, 2, etc.), select **waporipa** as the kernel:

### In VS Code
1. Open any `.ipynb` file
2. Click **Select Kernel** in the top-right of the notebook
3. Choose **Python Environments > waporipa**

### In Jupyter Notebook / JupyterLab
1. Open any `.ipynb` file
2. Go to **Kernel > Change Kernel**
3. Choose **waporipa**

> **Tip:** If `waporipa` does not appear in the kernel list, you may need to wait a moment or restart VS Code / Jupyter for it to detect the new environment.

### From the terminal
If you prefer to launch Jupyter from the command line:
```bash
conda activate waporipa
jupyter notebook
```
Your terminal prompt should now show `(waporipa)` at the beginning, indicating the environment is active.

# MAINTENANCE: Adding or updating dependencies

If you are working in the notebooks and need a new Python package to be installed, do the following:

---

### 1. Add the import to your notebook code

Use the package in your notebook as normal, e.g.:

```python
import seaborn as sns
```

---

### 2. Add the package to `environment.yml`

Open `environment.yml` (in the repository root) in a text editor and add the package name under the appropriate section. For example, to add `seaborn`:

  ```yaml
  dependencies:
    # ...existing packages...

    # Visualization
    - matplotlib
    - seaborn        # <-- add new packages here
  ```

> **Tip:** Only add packages you directly import — conda/mamba will automatically install any transitive dependencies. Prefer the conda-forge package name (search [anaconda.org](https://anaconda.org) if unsure).

---

### 3. Update the environment

Run **one** of the cells below to install the new package into the existing `waporipa` environment. The `--prune` flag also removes any packages no longer listed in the file.

#### Option A: mamba from this notebook (recommended)

````python
env_file = os.path.join(os.path.split(os.getcwd())[0], 'environment.yml')
!mamba env update -f {env_file} --prune
````

#### Option B: conda from this notebook

````python
env_file = os.path.join(os.path.split(os.getcwd())[0], 'environment.yml')
!conda env update -f {env_file} --prune
````

#### Option C: from the terminal

Copy and paste **one** of the following into your terminal (after activating the `waporipa` environment):

````python
# Using mamba
mamba env update -f environment.yml --prune

# Using conda
conda env update -f environment.yml --prune
````

### 4. After updating, **restart your Jupyter kernel** (Kernel > Restart) to pick up the changes.

## Useful commands reference

| Task | Command |
|---|---|
| Create the environment | `conda env create -f environment.yml` |
| Activate the environment | `conda activate waporipa` |
| Update after changes to environment.yml | `conda env update -f environment.yml --prune` |
| Deactivate the environment | `conda deactivate` |
| Remove the environment entirely | `conda env remove -n waporipa` |
| List all environments | `conda env list` |

Replace `conda` with `mamba` in any of the above for faster execution.
